## 3. Nạp đúng Kaggle Dataset canonical

Notebook ưu tiên mount `anhbodoidilinh08/canonical`, hỗ trợ cả layout `canonical/*.jsonl` và `*.jsonl` ở root, từ chối lựa chọn mơ hồ, kiểm tra kích thước/schema/ID leakage, và lưu SHA-256 provenance. Input: https://www.kaggle.com/datasets/anhbodoidilinh08/canonical


```python
KAGGLE_DATASET_SLUG = 'anhbodoidilinh08/canonical'
KAGGLE_DATASET_URL = f'https://www.kaggle.com/datasets/{KAGGLE_DATASET_SLUG}'
import kagglehub

# Resolve by owner/slug, never by a lookalike directory name. On Kaggle this
# reuses the attached input/cache; if import metadata was stripped, kagglehub
# downloads the declared dataset without altering other notebook inputs.
DATASET_MOUNT = Path(kagglehub.dataset_download(KAGGLE_DATASET_SLUG)).resolve()
SPLITS = ('train', 'validation', 'test')
EXPECTED_SPLIT_SIZES = {'train': 2000, 'validation': 400, 'test': 600}

def complete_split_dir(directory: Path) -> bool:
    return all((directory / f'{split}.jsonl').is_file() for split in SPLITS)

resolved_dirs = {
    train_path.parent.resolve()
    for train_path in DATASET_MOUNT.glob('**/train.jsonl')
    if complete_split_dir(train_path.parent)
}
assert len(resolved_dirs) == 1, (
    f'Expected one complete canonical split directory in {DATASET_MOUNT}, '
    f'found {[str(path) for path in sorted(resolved_dirs)]}'
)
CANONICAL_DIR = next(iter(resolved_dirs))
DATASET_SOURCE_MODE = 'kagglehub-owner-slug'

print('Dataset URL:', KAGGLE_DATASET_URL)
print('Dataset source mode:', DATASET_SOURCE_MODE)
print('Dataset mount/cache root:', DATASET_MOUNT)
print('Canonical split directory:', CANONICAL_DIR)

def read_jsonl(path: Path) -> list[dict]:
    records = []
    with path.open(encoding='utf-8') as handle:
        for line_number, line in enumerate(handle, start=1):
            if not line.strip():
                continue
            try:
                records.append(json.loads(line))
            except json.JSONDecodeError as exc:
                raise ValueError(f'Invalid JSON at {path}:{line_number}: {exc}') from exc
    return records

canonical = {split: read_jsonl(CANONICAL_DIR / f'{split}.jsonl') for split in SPLITS}
required_record_keys = {'id', 'scenario_family', 'input', 'expected'}
split_ids = {}
split_content_hashes = {}
split_message_hashes = {}
for split, records in canonical.items():
    assert len(records) == EXPECTED_SPLIT_SIZES[split], (
        f'{split} expected {EXPECTED_SPLIT_SIZES[split]} rows, found {len(records)}.'
    )
    assert all(required_record_keys.issubset(row) for row in records), f'{split} has missing record keys.'
    assert all(isinstance(row['input'], dict) and isinstance(row['expected'], dict) for row in records)
    ids = [str(row['id']) for row in records]
    assert len(ids) == len(set(ids)), f'{split} contains duplicate IDs.'
    split_ids[split] = set(ids)
    split_message_hashes[split] = {
        hashlib.sha256(str(row['input'].get('message') or '').casefold().strip().encode()).hexdigest()
        for row in records
    }
    split_content_hashes[split] = {
        hashlib.sha256(json.dumps(
            {'input': row['input'], 'expected': row['expected']},
            ensure_ascii=False, sort_keys=True, separators=(',', ':')
        ).encode()).hexdigest()
        for row in records
    }
    print(f'{split}: {len(records)} examples; families={len({row["scenario_family"] for row in records})}')

assert split_ids['train'].isdisjoint(split_ids['validation'])
assert split_ids['train'].isdisjoint(split_ids['test'])
assert split_ids['validation'].isdisjoint(split_ids['test'])
assert split_content_hashes['train'].isdisjoint(split_content_hashes['validation'])
assert split_content_hashes['train'].isdisjoint(split_content_hashes['test'])
assert split_content_hashes['validation'].isdisjoint(split_content_hashes['test'])
message_overlap_counts = {
    'train_validation': len(split_message_hashes['train'] & split_message_hashes['validation']),
    'train_test': len(split_message_hashes['train'] & split_message_hashes['test']),
    'validation_test': len(split_message_hashes['validation'] & split_message_hashes['test']),
}
print('Repeated latest-message hashes (diagnostic only):', message_overlap_counts)

def sha256_file(path: Path) -> str:
    digest = hashlib.sha256()
    with path.open('rb') as handle:
        for chunk in iter(lambda: handle.read(1024 * 1024), b''):
            digest.update(chunk)
    return digest.hexdigest()

dataset_manifest = {
    'dataset_slug': KAGGLE_DATASET_SLUG,
    'dataset_url': KAGGLE_DATASET_URL,
    'canonical_dir': str(CANONICAL_DIR),
    'source_mode': DATASET_SOURCE_MODE,
    'splits': {
        split: {
            'examples': len(canonical[split]),
            'sha256': sha256_file(CANONICAL_DIR / f'{split}.jsonl'),
        }
        for split in SPLITS
    },
}
(RUN_DIR / 'dataset_manifest.json').write_text(
    json.dumps(dataset_manifest, ensure_ascii=False, indent=2), encoding='utf-8'
)
print('Dataset integrity and split-leakage checks: PASS')
```

## 4. Chuyển canonical data sang ShareGPT

System prompt phải đến từ chính dataset canonical để không lệch production contract. Chỉ train và validation được chuyển; test giữ nguyên hoàn toàn cho held-out evaluation.


```python
prompt_candidates = [
    CANONICAL_DIR / 'openai' / 'train.jsonl',
    CANONICAL_DIR.parent / 'openai' / 'train.jsonl',
    DATASET_MOUNT / 'openai' / 'train.jsonl',
]
prompt_candidates = list(dict.fromkeys(path.resolve() for path in prompt_candidates))
openai_train = next((path for path in prompt_candidates if path.is_file()), None)
assert openai_train is not None, (
    'Canonical dataset is missing openai/train.jsonl, required to keep the training '
    f'system prompt aligned with production. Checked: {[str(path) for path in prompt_candidates]}'
)
openai_validation = openai_train.parent / 'validation.jsonl'
assert openai_validation.is_file(), 'Canonical dataset is missing openai/validation.jsonl.'
openai_sources = {'train': openai_train, 'validation': openai_validation}
prompt_values = set()
schema_values = {}
for split, source_path in openai_sources.items():
    source_rows = [json.loads(line) for line in source_path.read_text(encoding='utf-8').splitlines() if line.strip()]
    assert len(source_rows) == EXPECTED_SPLIT_SIZES[split]
    split_prompts = {
        row['messages'][0]['content']
        for row in source_rows
        if isinstance(row.get('messages'), list) and row['messages'] and row['messages'][0].get('role') == 'system'
    }
    assert len(split_prompts) == 1, f'{split} contains {len(split_prompts)} system prompts.'
    prompt_values.update(split_prompts)
    for row in source_rows:
        tools = row.get('tools') or []
        assert len(tools) == 1 and (tools[0].get('function') or {}).get('name') == 'AgentStructuredResult'
        schema = (tools[0].get('function') or {}).get('parameters')
        assert isinstance(schema, dict)
        schema_values[json.dumps(schema, sort_keys=True, separators=(',', ':'))] = schema
assert len(prompt_values) == 1, 'Train and validation system prompts differ.'
assert len(schema_values) == 1, 'Train and validation contain different AgentStructuredResult schemas.'
SYSTEM_PROMPT = next(iter(prompt_values))
AGENT_JSON_SCHEMA = next(iter(schema_values.values()))
(RUN_DIR / 'agent_structured_result.schema.json').write_text(
    json.dumps(AGENT_JSON_SCHEMA, ensure_ascii=False, indent=2), encoding='utf-8'
)
SYSTEM_PROMPT_SHA256 = hashlib.sha256(SYSTEM_PROMPT.encode()).hexdigest()
print('System prompt source:', openai_train)
print('System prompt length:', len(SYSTEM_PROMPT))
print('System prompt SHA-256:', SYSTEM_PROMPT_SHA256)

def to_sharegpt_row(record: dict) -> dict:
    return {
        'messages': [
            {'role': 'system', 'content': SYSTEM_PROMPT},
            {'role': 'user', 'content': json.dumps(record['input'], ensure_ascii=False, separators=(',', ':'))},
            {'role': 'assistant', 'content': json.dumps(record['expected'], ensure_ascii=False, separators=(',', ':'))},
        ]
    }

DATASET_DIR.mkdir(parents=True, exist_ok=True)
for split in ('train', 'validation'):
    rows = [to_sharegpt_row(record) for record in canonical[split]]
    output_path = DATASET_DIR / f'onbxrd_agent_{split}.json'
    output_path.write_text(json.dumps(rows, ensure_ascii=False), encoding='utf-8')
    print(f'Wrote {len(rows)} rows -> {output_path}')

message_tags = {
    'role_tag': 'role', 'content_tag': 'content',
    'user_tag': 'user', 'assistant_tag': 'assistant', 'system_tag': 'system',
}
dataset_info = {
    f'onbxrd_agent_{split}': {
        'file_name': f'onbxrd_agent_{split}.json',
        'formatting': 'sharegpt',
        'columns': {'messages': 'messages'},
        'tags': message_tags,
    }
    for split in ('train', 'validation')
}
(DATASET_DIR / 'dataset_info.json').write_text(
    json.dumps(dataset_info, ensure_ascii=False, indent=2), encoding='utf-8'
)
```

## 5. Kiểm tra schema và độ dài token

Cell kiểm tra role order, JSON payload, tokenizer chat template và tỷ lệ ví dụ vượt cutoff. Nếu hơn 5% bị cắt, notebook dừng thay vì âm thầm làm giảm chất lượng.


```python
from transformers import AutoTokenizer

for split in ('train', 'validation'):
    rows = json.loads((DATASET_DIR / f'onbxrd_agent_{split}.json').read_text(encoding='utf-8'))
    assert len(rows) == EXPECTED_SPLIT_SIZES[split]
    assert all([message['role'] for message in row['messages']] == ['system', 'user', 'assistant'] for row in rows)
    assert all(json.loads(row['messages'][1]['content']) for row in rows)
    assert all(json.loads(row['messages'][2]['content']) for row in rows)

from llamafactory.data.template import get_template_and_fix_tokenizer
from llamafactory.hparams import DataArguments
from jsonschema import Draft202012Validator, FormatChecker

contract_validator_for_data = Draft202012Validator(AGENT_JSON_SCHEMA, format_checker=FormatChecker())
target_validation_splits = SPLITS if EVAL_MODE == 'full' else ('train', 'validation')
for split in target_validation_splits:
    records = canonical[split]
    invalid_targets = [
        str(record['id']) for record in records
        if not contract_validator_for_data.is_valid(record['expected'])
    ]
    assert not invalid_targets, f'{split} expected outputs violate AgentStructuredResult schema: {invalid_targets[:10]}'

tokenizer_for_audit = AutoTokenizer.from_pretrained(
    BASE_MODEL, revision=MODEL_REVISION, token=HF_TOKEN, trust_remote_code=True
)
template_for_audit = get_template_and_fix_tokenizer(
    tokenizer_for_audit, DataArguments(template=LLAMA_TEMPLATE)
)
audit_lengths = []
for record in canonical['train'] + canonical['validation']:
    row = to_sharegpt_row(record)['messages']
    prompt_ids, response_ids = template_for_audit.encode_oneturn(
        tokenizer_for_audit,
        [row[1], row[2]],
        system=row[0]['content'],
    )
    audit_lengths.append(len(prompt_ids) + len(response_ids))

percentiles = {str(p): int(np.percentile(audit_lengths, p)) for p in (50, 90, 95, 99, 100)}
truncated = sum(length > CUTOFF_LEN for length in audit_lengths)
token_audit = {
    'cutoff_len': CUTOFF_LEN,
    'examples': len(audit_lengths),
    'length_percentiles': percentiles,
    'examples_over_cutoff': truncated,
    'fraction_over_cutoff': round(truncated / len(audit_lengths), 6),
}
(RUN_DIR / 'token_length_audit.json').write_text(json.dumps(token_audit, indent=2), encoding='utf-8')
print(json.dumps(token_audit, indent=2))
assert token_audit['fraction_over_cutoff'] <= 0.05, (
    'More than 5% of train/validation examples exceed CUTOFF_LEN. Increase the cutoff '
    'only after checking GPU headroom, or shorten redundant prompt/data fields.'
)
del tokenizer_for_audit
gc.collect()
print('ShareGPT schema and token-length audit: PASS')
```

## 6. Sinh cấu hình QLoRA có resume và best-checkpoint

Thiết lập dùng NF4 double quantization, FP16 cho T4, gradient checkpointing, SDPA, giới hạn checkpoint, chọn checkpoint validation tốt nhất, và tự resume checkpoint gần nhất.


```python
import yaml

def checkpoint_is_complete(path: Path) -> bool:
    required = ('trainer_state.json', 'optimizer.pt', 'scheduler.pt')
    return all((path / name).is_file() for name in required) and bool(list(path.glob('adapter_model*')))

checkpoint_dirs = sorted(
    [path for path in ADAPTER_DIR.glob('checkpoint-*') if checkpoint_is_complete(path)],
    key=lambda path: int(path.name.rsplit('-', 1)[-1]) if path.name.rsplit('-', 1)[-1].isdigit() else -1,
)
latest_checkpoint = checkpoint_dirs[-1] if checkpoint_dirs else None

train_config = {
    'model_name_or_path': BASE_MODEL,
    'model_revision': MODEL_REVISION,
    'template': LLAMA_TEMPLATE,
    'trust_remote_code': True,
    'stage': 'sft',
    'do_train': True,
    'finetuning_type': 'lora',
    'lora_target': LORA_TARGET,
    'lora_rank': LORA_RANK,
    'lora_alpha': LORA_ALPHA,
    'lora_dropout': 0.05,
    'quantization_bit': 4,
    'quantization_method': 'bnb',
    'quantization_type': 'nf4',
    'double_quantization': True,
    'low_cpu_mem_usage': True,
    'dataset': 'onbxrd_agent_train',
    'eval_dataset': 'onbxrd_agent_validation',
    'dataset_dir': str(DATASET_DIR),
    'cutoff_len': CUTOFF_LEN,
    'max_samples': EXPECTED_SPLIT_SIZES['train'],
    'overwrite_cache': True,
    'preprocessing_num_workers': 2,
    'dataloader_num_workers': 1,
    'output_dir': str(ADAPTER_DIR),
    'logging_steps': 10,
    'save_strategy': 'steps',
    'save_steps': 100,
    'save_total_limit': 2,
    'eval_strategy': 'steps',
    'eval_steps': 100,
    'load_best_model_at_end': True,
    'metric_for_best_model': 'eval_loss',
    'greater_is_better': False,
    'per_device_train_batch_size': 1,
    'per_device_eval_batch_size': 1,
    'gradient_accumulation_steps': GRADIENT_ACCUMULATION_STEPS,
    'eval_accumulation_steps': 1,
    'learning_rate': LEARNING_RATE,
    'optim': OPTIM_ALGO,
    'num_train_epochs': 3.0,
    'lr_scheduler_type': 'cosine',
    'warmup_ratio': 0.05,
    'max_grad_norm': 1.0,
    'fp16': True,
    'bf16': False,
    'disable_gradient_checkpointing': False,
    'use_reentrant_gc': True,
    'flash_attn': TRAIN_ATTENTION,
    'plot_loss': True,
    'report_to': 'none',
    'seed': SEED,
    'data_seed': SEED,
    'save_safetensors': True,
    'include_num_input_tokens_seen': True,
    'overwrite_output_dir': latest_checkpoint is None,
}
training_config_for_identity = {
    key: value for key, value in train_config.items()
    if key not in {'resume_from_checkpoint', 'overwrite_output_dir'}
}
training_identity = {
    'effective_train_config': training_config_for_identity,
    'dataset_hashes': dataset_manifest['splits'],
    'system_prompt_sha256': SYSTEM_PROMPT_SHA256,
    'llamafactory_commit': LLAMA_FACTORY_COMMIT,
    'runtime_versions_sha256': sha256_file(RUN_DIR / 'runtime_versions.json'),
}
training_identity_hash = hashlib.sha256(
    json.dumps(training_identity, sort_keys=True).encode()
).hexdigest()
training_manifest_path = RUN_DIR / 'training_identity.json'
if latest_checkpoint is not None:
    assert training_manifest_path.is_file(), 'Checkpoint exists without training_identity.json; use a new RUN_NAME.'
    previous_identity = json.loads(training_manifest_path.read_text(encoding='utf-8'))
    assert previous_identity.get('sha256') == training_identity_hash, (
        'Existing checkpoint belongs to a different dataset/model/config/runtime. Use a new RUN_NAME.'
    )
    train_config['resume_from_checkpoint'] = str(latest_checkpoint)
    print('Training will resume from:', latest_checkpoint)
training_manifest_path.write_text(json.dumps({
    'sha256': training_identity_hash, 'config': training_identity
}, ensure_ascii=False, indent=2), encoding='utf-8')

config_path = RUN_DIR / 'train_config.yaml'
config_path.write_text(yaml.safe_dump(train_config, sort_keys=False, allow_unicode=True), encoding='utf-8')
print(config_path.read_text(encoding='utf-8'))
```

## 9. Inference 4-bit an toàn bộ nhớ và smoke test

Model luôn map vào GPU 0, có giới hạn VRAM, giải phóng cache trước load, bật KV cache chỉ lúc inference, dùng greedy decoding và báo lỗi OOM có thông tin.


```python
from peft import PeftModel
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig

def release_cuda(model=None) -> None:
    if model is not None:
        try:
            model.cpu()
        except Exception:
            pass
        del model
    gc.collect()
    torch.cuda.empty_cache()
    torch.cuda.ipc_collect()

release_cuda()
quantization_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type='nf4',
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True,
)
tokenizer = AutoTokenizer.from_pretrained(BASE_MODEL, revision=MODEL_REVISION, token=HF_TOKEN, trust_remote_code=True)
if tokenizer.pad_token_id is None:
    tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = 'left'

def load_for_inference(adapter_path: Path | None = None):
    release_cuda()
    base = AutoModelForCausalLM.from_pretrained(
        BASE_MODEL,
        token=HF_TOKEN,
        revision=MODEL_REVISION,
        quantization_config=quantization_config,
        device_map={'': 0},
        max_memory={0: f'{INFERENCE_MAX_MEMORY_GIB}GiB'},
        torch_dtype=torch.float16,
        low_cpu_mem_usage=True,
        attn_implementation='eager' if LLAMA_TEMPLATE == 'gemma2' else 'sdpa',
        trust_remote_code=True,
    )
    model = PeftModel.from_pretrained(base, adapter_path, is_trainable=False) if adapter_path else base
    model.eval()
    model.config.use_cache = True
    return model

def render_prompt(context: dict) -> dict:
    return {'role': 'user', 'content': json.dumps(context, ensure_ascii=False, separators=(',', ':'))}

def build_inference_messages(context: dict) -> list[dict]:
    user_message = render_prompt(context)
    # Gemma and Mistral tokenizers commonly reject a separate system role.
    # Folding it into the first user turn preserves the instruction while
    # remaining compatible with their instruction templates.
    if LLAMA_TEMPLATE in {'gemma', 'gemma2', 'mistral'}:
        return [{
            'role': 'user',
            'content': SYSTEM_PROMPT + '\n\n' + user_message['content'],
        }]
    return [{'role': 'system', 'content': SYSTEM_PROMPT}, user_message]

@torch.inference_mode()
def generate_agent_json(inference_model, context: dict, max_new_tokens: int = EVAL_MAX_NEW_TOKENS) -> str:
    messages = build_inference_messages(context)
    template_kwargs = dict(
        conversation=messages, add_generation_prompt=True, tokenize=True, return_tensors='pt'
    )
    try:
        inputs = tokenizer.apply_chat_template(enable_thinking=False, **template_kwargs)
    except TypeError:
        inputs = tokenizer.apply_chat_template(**template_kwargs)
    inputs = inputs.to('cuda:0')
    try:
        output_ids = inference_model.generate(
            inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            use_cache=True,
            pad_token_id=tokenizer.pad_token_id,
            eos_token_id=tokenizer.eos_token_id,
        )
    except torch.OutOfMemoryError as exc:
        torch.cuda.empty_cache()
        raise RuntimeError(
            f'Inference OOM at input_tokens={inputs.shape[-1]}, max_new_tokens={max_new_tokens}. '
            'Reduce EVAL_MAX_NEW_TOKENS before reducing evaluation coverage.'
        ) from exc
    decoded = tokenizer.decode(output_ids[0, inputs.shape[-1]:], skip_special_tokens=True).strip()
    if 'deepseek' in BASE_MODEL.lower() and '</think>' in decoded:
        decoded = decoded.rsplit('</think>', 1)[-1].strip()
    return decoded

development_records = canonical['validation'] if EVAL_MODE == 'smoke' else canonical['test']
finetuned_model = load_for_inference(ADAPTER_DIR)
print('Fine-tuned adapter loaded; allocated GiB:', round(torch.cuda.memory_allocated() / 1024**3, 2))
for record in (development_records[0], development_records[len(development_records) // 2]):
    raw = generate_agent_json(finetuned_model, record['input'])
    print('SCENARIO:', record['scenario_family'])
    print('USER:', record['input'].get('message'))
    print('MODEL OUTPUT:', raw[:1000])
    print('=' * 80)
```

## 10. Evaluation contract có thể resume

Mặc định `EVAL_MODE='smoke'` đánh giá 120 case cân bằng từ **validation**, nên test 600 case vẫn được giữ kín cho nghiệm thu. Chỉ `EVAL_MODE='full'` mới dùng toàn bộ test set. Output phải là đúng một JSON contract; checkpoint đánh giá chỉ được reuse khi model, runtime, prompt, schema, generation config và case IDs đều khớp.


```python
assert EVAL_MODE in {'smoke', 'full'}
assert 1 <= SMOKE_EVAL_CASES <= len(canonical['test'])
from jsonschema import Draft202012Validator, FormatChecker
contract_validator = Draft202012Validator(AGENT_JSON_SCHEMA, format_checker=FormatChecker())
EVALUATOR_VERSION = 'onbxrd-contract-eval-v3'
VALID_INTENTS = {
    'handbook_question', 'ticket_question', 'create_ticket', 'clarification_response',
    'create_announcement', 'announcement_edit', 'publish_announcement',
}

def stable_rank(record: dict) -> str:
    payload = f'{SEED}:{record["scenario_family"]}:{record["id"]}'.encode()
    return hashlib.sha256(payload).hexdigest()

def select_eval_records(records: list[dict], limit: int | None) -> list[dict]:
    ordered = sorted(records, key=lambda row: (row['scenario_family'], stable_rank(row)))
    if limit is None or limit >= len(ordered):
        return ordered
    by_family = defaultdict(list)
    for row in ordered:
        by_family[row['scenario_family']].append(row)
    selected = []
    family_names = sorted(by_family)
    while len(selected) < limit and any(by_family.values()):
        for family in family_names:
            if by_family[family] and len(selected) < limit:
                selected.append(by_family[family].pop(0))
    return selected

evaluation_split = 'test' if EVAL_MODE == 'full' else 'validation'
eval_limit = None if EVAL_MODE == 'full' else SMOKE_EVAL_CASES
test_records = select_eval_records(canonical[evaluation_split], eval_limit)
selection_ids = [str(row['id']) for row in test_records]
(RUN_DIR / 'evaluation_case_ids.json').write_text(json.dumps(selection_ids, indent=2), encoding='utf-8')
print(f'Evaluation mode={EVAL_MODE}; split={evaluation_split}; cases={len(test_records)}; families={len({row["scenario_family"] for row in test_records})}')

def normalized_items(values: list) -> list[str]:
    return sorted(json.dumps(value, ensure_ascii=False, sort_keys=True) for value in values)

def is_schema_valid(value: dict | None) -> bool:
    return (
        isinstance(value, dict)
        and contract_validator.is_valid(value)
        and isinstance(value.get('assistant_message'), str)
        and bool(value['assistant_message'].strip())
    )

def parse_contract_json(raw: str) -> dict | None:
    normalized = raw.strip()
    try:
        value = json.loads(normalized)
    except json.JSONDecodeError:
        return None
    return value if is_schema_valid(value) else None

CHECK_NAMES = {
    'schema_valid', 'assistant_message_present', 'intent_match', 'action_match',
    'action_contract_match', 'draft_contract_match', 'missing_fields_exact',
    'required_fields_present', 'references_match', 'role_boundary',
    'publish_confirmation_boundary',
}
TICKET_FIELDS = {'unit_id', 'category', 'priority', 'title', 'description', 'location'}
ANNOUNCEMENT_FIELDS = {'title', 'content', 'scopes'}
ANNOUNCEMENT_ACTIONS = {'create_announcement_draft', 'update_announcement', 'publish_announcement'}

def evaluate_case(record: dict, inference_model) -> dict:
    raw = generate_agent_json(inference_model, record['input'])
    actual = parse_contract_json(raw)
    expected = record['expected']
    expected_action = expected.get('action') or {}
    expected_draft = expected.get('draft') or {}
    schema_valid = actual is not None
    actual_action = actual.get('action') or {} if schema_valid else {}
    actual_draft = actual.get('draft') or {} if schema_valid else {}
    actual_missing = set(actual.get('missing_fields') or []) if schema_valid else set()
    expected_missing = set(expected.get('missing_fields') or [])
    actor_role = record['input'].get('actor_role')
    role_known = actor_role in {'resident', 'management'}
    role_boundary = schema_valid and role_known and not (
        actor_role == 'resident' and actual_action.get('type') in ANNOUNCEMENT_ACTIONS
    ) and not (
        actor_role == 'management' and actual_action.get('type') == 'create_ticket'
    )
    known_expected_fields = {
        key for key, value in expected_draft.items() if value not in (None, '', [])
    }
    required_fields = (
        TICKET_FIELDS if expected_action.get('type') == 'create_ticket'
        else ANNOUNCEMENT_FIELDS if expected_action.get('type') in ANNOUNCEMENT_ACTIONS
        else known_expected_fields
    )
    checks = {
        'schema_valid': schema_valid,
        'assistant_message_present': schema_valid and bool(actual['assistant_message'].strip()),
        'intent_match': schema_valid and actual.get('intent') == expected.get('intent'),
        'action_match': schema_valid and actual_action.get('type') == expected_action.get('type'),
        'action_contract_match': schema_valid and actual_action == expected_action,
        'draft_contract_match': schema_valid and actual_draft == expected_draft,
        'missing_fields_exact': schema_valid and actual_missing == expected_missing,
        'required_fields_present': schema_valid and all(
            actual_draft.get(field) not in (None, '', [])
            for field in required_fields - expected_missing
        ),
        'references_match': schema_valid and normalized_items(actual.get('attachment_ids') or []) == normalized_items(expected.get('attachment_ids') or []) and normalized_items(actual.get('citations') or []) == normalized_items(expected.get('citations') or []),
        'role_boundary': role_boundary,
        'publish_confirmation_boundary': schema_valid and (
            actual_action.get('type') != 'publish_announcement'
            or expected_action.get('type') == 'publish_announcement'
        ),
    }
    return {
        'id': str(record['id']), 'family': record['scenario_family'],
        'raw': raw, 'actual': actual, 'expected': expected,
        'checks': checks,
        'quality_score': round(100 * sum(checks.values()) / len(checks), 1),
        'all_checks_pass': all(checks.values()),
    }

def atomic_json_write(path: Path, payload) -> None:
    temporary = path.with_suffix(path.suffix + '.tmp')
    temporary.write_text(json.dumps(payload, ensure_ascii=False, indent=2), encoding='utf-8')
    temporary.replace(path)

def hash_files(paths: list[Path]) -> str:
    digest = hashlib.sha256()
    for path in sorted(paths, key=str):
        digest.update(str(path.relative_to(ADAPTER_DIR)).encode())
        digest.update(sha256_file(path).encode())
    return digest.hexdigest()

def adapter_fingerprint() -> str:
    candidates = [
        path for path in ADAPTER_DIR.rglob('*')
        if path.is_file() and (
            path.name.startswith('adapter_model') or path.name == 'adapter_config.json'
        ) and 'checkpoint-' not in path.parts
    ]
    assert candidates, 'Final adapter weights/config are missing from ADAPTER_DIR.'
    return hashlib.sha256(
        (BASE_MODEL + ':' + MODEL_REVISION + ':' + hash_files(candidates)).encode()
    ).hexdigest()

evaluation_config = {
    'evaluator_version': EVALUATOR_VERSION,
    'system_prompt_sha256': hashlib.sha256(SYSTEM_PROMPT.encode()).hexdigest(),
    'template': LLAMA_TEMPLATE,
    'model_revision': MODEL_REVISION,
    'max_new_tokens': EVAL_MAX_NEW_TOKENS,
    'eval_mode': EVAL_MODE,
    'evaluation_split': evaluation_split,
    'case_ids': selection_ids,
    'schema_sha256': hashlib.sha256(json.dumps(AGENT_JSON_SCHEMA, sort_keys=True).encode()).hexdigest(),
    'quantization': {'type': 'nf4', 'double_quantization': True, 'compute_dtype': 'float16'},
    'attention': 'eager' if LLAMA_TEMPLATE == 'gemma2' else 'sdpa',
    'max_memory_gib': INFERENCE_MAX_MEMORY_GIB,
    'runtime_versions_sha256': sha256_file(RUN_DIR / 'runtime_versions.json'),
}
evaluation_config_hash = hashlib.sha256(
    json.dumps(evaluation_config, sort_keys=True).encode()
).hexdigest()

def run_evaluation(label: str, inference_model, records: list[dict], fingerprint: str) -> list[dict]:
    path = RUN_DIR / f'{label}_evaluation_results.json'
    manifest = {
        'model': BASE_MODEL,
        'fingerprint': fingerprint,
        'evaluation_config_hash': evaluation_config_hash,
        'case_ids': selection_ids,
        'results': [],
    }
    if RESUME_EVALUATION and path.exists():
        try:
            previous = json.loads(path.read_text(encoding='utf-8'))
            previous_results = previous.get('results')
            previous_rows_valid = isinstance(previous_results, list) and all(
                isinstance(row, dict)
                and isinstance(row.get('id'), str)
                and isinstance(row.get('checks'), dict)
                and set(row['checks']) == CHECK_NAMES
                and isinstance(row.get('quality_score'), (int, float))
                and isinstance(row.get('all_checks_pass'), bool)
                for row in (previous_results or [])
            )
            previous_ids = [row['id'] for row in previous_results] if previous_rows_valid else []
            compatible = all(previous.get(key) == manifest[key] for key in (
                'model', 'fingerprint', 'evaluation_config_hash', 'case_ids'
            )) and len(previous_ids) == len(set(previous_ids)) and set(previous_ids).issubset(selection_ids)
            if compatible:
                manifest = previous
                print(f'Resuming {label}: {len(manifest["results"])}/{len(records)} complete')
            else:
                print(f'Ignoring incompatible {label} evaluation checkpoint.')
        except (OSError, ValueError, TypeError, json.JSONDecodeError) as exc:
            print(f'Ignoring unreadable {label} evaluation checkpoint: {exc}')
    completed = {row['id'] for row in manifest['results']}
    for index, record in enumerate(records, start=1):
        if str(record['id']) not in completed:
            manifest['results'].append(evaluate_case(record, inference_model))
            atomic_json_write(path, manifest)
        if index % 10 == 0 or index == len(records):
            print(f'{label}: {index}/{len(records)}')
    result_by_id = {row['id']: row for row in manifest['results']}
    assert all(case_id in result_by_id for case_id in selection_ids)
    return [result_by_id[case_id] for case_id in selection_ids]

def summarize_results(evaluation_results: list[dict]) -> dict:
    assert evaluation_results, 'Cannot summarize an empty evaluation.'
    def aggregate(rows: list[dict]) -> dict:
        return {
            'examples': len(rows),
            **{
                f'{check}_rate': round(sum(row['checks'][check] for row in rows) / len(rows), 4)
                for check in rows[0]['checks']
            },
            'average_quality_score': round(sum(row['quality_score'] for row in rows) / len(rows), 2),
            'all_checks_pass_rate': round(sum(row['all_checks_pass'] for row in rows) / len(rows), 4),
        }
    overall = aggregate(evaluation_results)
    by_family = defaultdict(list)
    for row in evaluation_results:
        by_family[row['family']].append(row)
    summary = dict(overall)
    summary['evaluation_mode'] = EVAL_MODE
    summary['evaluator_version'] = EVALUATOR_VERSION
    summary['evaluation_config_hash'] = evaluation_config_hash
    summary['case_ids_sha256'] = hashlib.sha256('\n'.join(selection_ids).encode()).hexdigest()
    summary['per_family'] = {family: aggregate(rows) for family, rows in sorted(by_family.items())}
    return summary

results = run_evaluation('finetuned', finetuned_model, test_records, adapter_fingerprint())
metrics = summarize_results(results)
atomic_json_write(RUN_DIR / 'evaluation_summary.json', metrics)
print(json.dumps({key: value for key, value in metrics.items() if key != 'per_family'}, indent=2))
```

Adapter được giải phóng hoàn toàn trước khi load base model. Hai model dùng đúng cùng thứ tự case; báo cáo chứa delta tổng thể và breakdown theo scenario family.
